```python
class Car:
    def __init__ (self, model, year, country):
        self.model = model
        self.year = year
        self.country = country

    def paint(self,color):
        print( f"the car is painted {color}.")

    def __str__(self):
        return f"{self.model} {self.year} {self.country}"

    def __repr__(self): # ideally looks like the code that would recreate an object
        # used for debugging purpose/ developers
        return f"Car({self.model!r} {self.year!r} {self.country!r})"

car1 = Car("Audi", 2001,"Germany")
print(car1)

print(car1.model)
print(car1.year)
print(car1.country)
car1.paint("Blue")

print([car1])
car1
```

```python
# Parameter order : positional, default, keywords only
# *args - collect extra positional arguments into tuple
# **kwargs - collect extra keyword arguments into dict

def student (name,age = 20 ,*, department, stream):
    print (f"{name} is {age} years old.")
    print(f"{name} is {department} department and {stream} stream.")

student("Abebe", department="Software", stream="AI")

print("\n")

def describe (**fields):
    for key, val in fields.items():
        print(f"{key} : {val}")

describe(number = 50, year = 2020 , department = "Software Engineer")
```

```python
# mutable default arguments

def log(event,history=[]): # the default argument is shared throughout the function calls
    history.append(event)
    print(event , history)

log("Birthday")
log("Graduation")

# fix
def log2 (*event,history=None):
    if history is None:
        history = []
    for e in event:
        history.append(e)
    print(event, history)

log2 ("birthday", "Wedding")
log2 ("graduation")
```

```python
# Classes
class Account:
    currency = "ETB"
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        return self.balance

    def withdraw(self, amount):
        if self.balance < amount:
            raise ValueError("Insufficient funds")

        self.balance = self.balance - amount
        return self.balance

    def __repr__(self):
        return f"{type(self).__name__}(balance={self.balance!r})"

    #type(self).__name__ : pick up the real class name automatically

    def __str__(self):
        return f"This account has {self.balance} {self.currency}"

acc = Account(100)
print(acc)
print(repr(acc))

class SavingAccount(Account):
    def __init__(self, balance, rate):
        super().__init__(balance)
        self.rate = rate

    def __str__(self):
        return f"{super().__str__()} and {self.rate}% interest rate"

acc2 = SavingAccount(100, rate=15)
d = acc2.deposit
d(200)
acc2.withdraw(50)
print(acc2)
```

```python
# Dataclasses
import json
from dataclasses import dataclass, field, asdict, replace, FrozenInstanceError


@dataclass
class Expense:
    amount: int
    category: str
    note: str=""
    tags: list= field(default_factory=list)

@dataclass(frozen=True)
class Budget:
    category: str
    limit: int

budget = Budget("Finance", 4000)
try:
    budget.limit = 3000
except FrozenInstanceError as e:
    print("can't change the value", e)

new_budget=replace(budget, limit=5000)
print(budget)
print(new_budget, end="\n\n")

exp1 = Expense(amount=100, category="Food", note="Food expense", tags=["Papaya","Avocado"])
exp2 = Expense(amount=200, category="School", note="School expense")
exp3 = Expense(amount=300, category="Medical", note="Annual checkup expense")

expenses = [exp1, exp2, exp3]
dicts = [asdict(e) for e in expenses]
json_data=json.dumps(dicts)
unpacked = [Expense(**e) for e in json.loads(json_data)]

print(expenses, end="\n\n")
print(dicts, end="\n\n")
print(json_data, end="\n\n")
print(unpacked)
print(expenses == unpacked)

```

```python
# Exceptions
class InvalidAmountError(Exception):
    """Amount is not a positive number"""
def parse_amount(text):
    try:
        amount = float(text)
    except ValueError as e:
        raise InvalidAmountError(f"Not a number: {text!r}") from e

    if amount <= 0:
        raise InvalidAmountError(f"Not a positive number: {text!r}")

    return amount

parse_amount("12.5")


def process_all(numbers):
    total = 0
    processed = 0
    try:
        for i in numbers:
            processed+=1
            try:
                total+=parse_amount(i)
            except InvalidAmountError as e:
                print("Error:",e)
        return total
    finally:
        print(f"{processed} inputs are processed")

numbers = [10, "abc", -0.3, 7.5]
process_all(numbers); # ; hides the auto display of the last line in jupyter
```
