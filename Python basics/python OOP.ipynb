```python
class Car:
    def __init__ (self, model, year, country):
        self.model = model
        self.year = year
        self.country = country

    def paint(self,color):
        print( f"the car is painted {color}.")

    def __str__(self):
        return f"{self.model} {self.year} {self.country}"

    def __repr__(self): # ideally looks like the code that would recreate an object
        # used for debugging purpose/ developers
        return f"Car({self.model!r} {self.year!r} {self.country!r})"

car1 = Car("Audi", 2001,"Germany")
print(car1)

print(car1.model)
print(car1.year)
print(car1.country)
car1.paint("Blue")

print([car1])
car1
```

```python
# generators -- good for memory efficiency
def count(n):
    for i in range(n):
        yield i

num = count(3)
print(next(num))
print(next(num))
```

```python
# Parameter order : positional, default, keywords only
# *args - collect extra positional arguments into tuple
# **kwargs - collect extra keyword arguments into dict

def student (name,age = 20 ,*, department, stream):
    print (f"{name} is {age} years old.")
    print(f"{name} is {department} department and {stream} stream.")

student("Abebe", department="Software", stream="AI")

print("\n")

def describe (**fields):
    for key, val in fields.items():
        print(f"{key} : {val}")

describe(number = 50, year = 2020 , department = "Software Engineer")
```

```python
# arguments
def summarize(*amounts, currency="ETB"):
    return len(amounts), sum(amounts), sum(amounts)/len(amounts), currency

count, total, average, currency = summarize(50,38,78,currency="USD")
print(f"{count} expenses, total {total} {currency}.")
```

```python
# mutable default arguments

def log(event,history=[]): # the default argument is shared throughout the function calls
    history.append(event)
    print(event , history)

log("Birthday")
log("Graduation")

# fix
def log2 (*event,history=None):
    if history is None:
        history = []
    for e in event:
        history.append(e)
    print(event, history)

log2 ("birthday", "Wedding")
log2 ("graduation")
```
